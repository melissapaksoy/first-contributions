```python
# a
# Author: Melissa Paksoy
# Assignment: #1

#b
gym_member = "Alex Alliton"   #string
preferred_weight_kg = 20.5    #float
highest_reps = 25             #integer
membership_active = True      #boolean

#c
# Dictionary where keys are names of friends, and values are tuples of integers representing workout minutes for different activities
workout_stats = {
    "Alex": (30, 45, 20),
    "Tim": (25, 60, 15),
    "Laura": (75, 10, 25)
}

#d
# Loop through each friend in the dictionary and calculate total workout minutes
for friend, minutes in list(workout_stats.items()):  # Use list() to avoid modifying the dictionary while iterating
    total_minutes = sum(minutes)  # Calculate total workout minutes
    workout_stats[f"{friend}_Total"] = total_minutes  # Add new key-value pair

#e
# 2D list where each row represents a friend, and each column represents one activity (yoga, running, weightlifting)
workout_list = [list(minutes) for friend, minutes in workout_stats.items() if not friend.endswith("_Total")]

#f
# Slice to extract and print minutes for yoga and running for all friends
allmins_yoga_running = [row[:2] for row in workout_list]
print("Yoga and Running Minutes:", allmins_yoga_running)

# Slice to extract and print minutes for weightlifting for the last two friends
weightlifting_lasttwof = [row[2] for row in workout_list[-2:]]
print("Weightlifting Minutes (Last Two Friends):", weightlifting_lasttwof)

#g
# Check if any friend's total workout minutes are greater than or equal to 120
for friend, minutes in workout_stats.items():
    if friend.endswith("_Total") and minutes > 120:  # Only check total values
        name = friend.split("_")[0]  # Extract name
        print(f"Great job staying active, {name}!")
#h
# Allow user to input a friend's name and display their workout minutes
entered_name = input("Enter a friend's name: ")

if entered_name in workout_stats:
    workout_data = workout_stats[entered_name]
    total_minutes = workout_stats.get(f"{entered_name}_Total", "N/A")
    print(f"\n{entered_name}'s Workout Minutes:")
    print(f"Yoga: {workout_data[0]} min")
    print(f"Running: {workout_data[1]} min")
    print(f"Weightlifting: {workout_data[2]} min")
    print(f"Total: {total_minutes} min")
else:
    print(f"Friend {entered_name} not found in the records.")

#i
# Find the friend with the highest and lowest total workout minutes
totals = {friend: minutes for friend, minutes in workout_stats.items() if friend.endswith("_Total")}

#Print the Highest Workout Friend's Name & Minutes
highest_friend = max(totals, key=totals.get)
print(f"\nFriend with the highest total workout minutes: {highest_friend.replace('_Total', '')} ({totals[highest_friend]} min)")

#Print the Lowest Workout Friend's Name & Minutes
lowest_friend = min(totals, key=totals.get)
print(f"Friend with the lowest total workout minutes: {lowest_friend.replace('_Total', '')} ({totals[lowest_friend]} min)")
```
